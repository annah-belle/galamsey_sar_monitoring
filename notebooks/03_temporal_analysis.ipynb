# 03 — Temporal Analysis: 2015–2025

This notebook presents the annual monitoring results within the **Common Monitoring Footprint (CMF)**, defined as the spatial intersection of June Sentinel-1 coverage from 2015–2025.

The monitoring workflow produces annual model probabilities and applies the operational threshold of **0.70** to derive detected areas.

```python
import pandas as pd
import matplotlib.pyplot as plt

annual = pd.DataFrame({
    "Year": list(range(2015, 2026)),
    "Mean_probability": [0.140, 0.185, 0.205, 0.247, 0.239, 0.241, 0.240, 0.220, 0.202, 0.240, 0.255],
    "Detected_area_km2": [298.1, 695.0, 524.8, 723.7, 607.7, 638.2, 648.8, 610.9, 548.1, 761.1, 755.7],
})

display(annual)
```

```python
fig, ax = plt.subplots(figsize=(10, 5))
ax.plot(annual["Year"], annual["Mean_probability"], marker="o")
ax.set_xlabel("Year")
ax.set_ylabel("Mean galamsey probability")
ax.set_title("Annual mean model probability within the CMF")
ax.grid(alpha=0.25)
plt.tight_layout()
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(10, 5))
ax.bar(annual["Year"], annual["Detected_area_km2"])
ax.set_xlabel("Year")
ax.set_ylabel("Detected area (km²)")
ax.set_title("Annual detected area at the operational probability threshold")
plt.tight_layout()
plt.show()
```

## Reading the time series

The documented annual results show variation in both mean model probability and detected area across the 2015–2025 period.

The largest documented detected area is **761.1 km² in 2024**, followed by **755.7 km² in 2025**.

These values should be interpreted as model-derived detections within the CMF and should not be described as direct measurements of the true physical extent of mining without independent validation.
